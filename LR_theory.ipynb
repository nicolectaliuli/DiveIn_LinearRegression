What's Linear Regression?

Linear Regression is math approach that fits a linear function over a data distribution through the minimization of sum of the squared error (MSE) between the real data and the equation value. On Machine Learning we use this approach to find a linear relation among a scalar target and some exploratory features. The equation we are looking for is known as Linear Equation represented by the following structure:

_Linear Equation_


$f(x) = \sum_{i=0}^{N} a_ix_i$


The equation responsable for "finding" the best fit, is called _Loss Function (LF)_, responsible to quantifies the error (vertical distance) of the predicted value over the real values:

_Loss Function_

$L(m,b) = \frac{1}{n}\sum_{i=0}^{N}[y_i - (mx_i-b)]^2$

where
-   $y_i$: is the real observed value
-   $m*x_i-b$: is the predicted value


How does Linear Regression Models works?

The method behind the model is conducted by the Gradient Descent optimization, the ideia is observe the error vairation in *LF* in respect to smal variation in *m* and *b*. To do that, we are going to apply a partial derivation on *LF* in the both variables, intercept (*b*) and slope (*m*). This allow us to have a vector pointing the direction of greatest error increase (not convinced? see the notes).

Partial derivative in respect of *m*

$\frac{\partial L}{\partial m} = \frac{-2}{n}\sum_{i=0}^{N} x_n*[y_i - (mx_i - b)]$

Partial derivative in respect of *b*

$\frac{\partial L}{\partial b} = \frac{-2}{n}\sum_{i=0}^{N} y_i - (mx_i - b)$

The vector with both variation will be:

$\partial L = \frac{\partial L}{\partial m} \partial m + \frac{\partial L}{\partial b} \partial b =\begin{pmatrix}
  \frac{\partial L}{\partial m}\\
  \frac{\partial L}{\partial b}
\end{pmatrix}  \begin{pmatrix}
  \partial m\\
  \partial b
\end{pmatrix}= \nabla L(m,b) \cdot \partial\vec{p}$

Found our Gradient Descent, if we want to minimize the error, we are going to follow the inverse direction of greates variation. We now have the direction and the variables we need to optimize, but how? The ideia is to start with both as random values (zero, to make our lives easier), and update their values each leraning round as we call *Epoch* and how much we will consider the Gradient direction, *Learning Rates*. To continue in the next step we need to understand how *Learning Rates (LR)* and *Epochs* works.

Imagine that our variables, starting with zero, are going to be updated as:

$m_{i+1} = m_i - \alpha \nabla L(m,b)$

$b_{i+1} = b_i - \alpha \nabla L(m,b)$,

where the updated variable is the pevious one, minus our *LR* times Gradient Descent

If every time we update our variables we give too much credit to the decrease direction, we can have a quickly convergence but we have a chance to diverge our result. In other hand if we don't give credit enough our convergence can be very slow. This credit we give in the given direction by $\nabla L(m,b)$ is called Learning Rate and is symbolized by $\alpha$

Now that we have the whole structure we need a frequency of update, that where our *Epoch* comes. In general we have three formats

- Batch Gradient Descent: here we update  $\nabla L(m,b)$ using all data avaliable. It's a safe step, we go all over the data, take the result vector as direction
- Stochastic Gradient Descent: here we update  $\nabla L(m,b)$ for every point in our data. It's a chaotic step, we take a step every point
- Mini Batch Gradient Descent: here we update  $\nabla L(m,b)$ for every N/batch in our data. It's a moderate step, we take a step for each group of data we observe.

In the completed logic we have:

Our GD giving us a direction to the gratest decrease point, our LR give us how big is our step in this direction and our epoch give us how many steps we are going to take and based on how many data.

NOTES: 

1) *Gradient Descent*

to max the variation,$||d\vec{p}||$, must to be 1, so we capture all the variation, using Cauchy-Schwarz inequality we have:

$\nabla L \cdot d\vec{p} \leq ||\nabla L||\cdot ||d\vec{p}||$

following with $||d\vec{p}||$ = 1, then

$\vec{\nabla} L \cdot d\vec{p} \leq ||\nabla L||$, then


$d\vec{p} \leq \frac{\vec{\nabla} L}{||\nabla L||}$

so we see now that to follow the greatest increase direction, we must follow the same direction as the Gradient $\vec{\nabla} L(m,b)$

O que é Regressão Linear?

A Regressão Linear é uma abordagem matemática que ajusta uma função linear a uma distribuição de dados, minimizando a soma dos quadrados dos erros (MSE) entre os dados reais e o valor da equação. Em Aprendizado de Máquina, usamos essa abordagem para encontrar uma relação linear entre uma variável escalar alvo e algumas características exploratórias. A equação que procuramos é conhecida como Função Afim e representada pela seguinte estrutura:

_Função Afim_


$f(x) = \sum_{i=0}^{N} a_ix_i$


A equação responsável por "encontrar" o melhor ajuste é chamada de _Função de Perda (LP)_, que quantifica o erro (distância vertical) do valor previsto em relação aos valores reais:

_Função de Perda_

$L(m,b) = \frac{1}{n}\sum_{i=0}^{N}[y_i - (mx_i-b)]^2$

onde:
- $y_i$: é o valor real observado
- $m*x_i-b$: é o valor previsto


Como funcionam os modelos de regressão linear?

O método por trás do modelo é conduzido pela otimização por Gradiente Descendente. A ideia é observar a variação do erro em *LF* em relação a pequenas variações em *m* e *b*. Para isso, aplicaremos uma derivada parcial em *LF* em relação às duas variáveis, intercepto (*b*) e inclinação (*m*). Isso nos permite obter um vetor que aponta a direção do maior aumento do erro (ainda não está convencido? Veja as notas).

Derivada parcial em relação a *m*

$\frac{\partial L}{\partial m} = \frac{-2}{n}\sum_{i=0}^{N} x_n*[y_i - (mx_i - b)]$

Derivada Parcial em relação a *b*

$\frac{\partial L}{\partial b} = \frac{-2}{n}\sum_{i=0}^{N} y_i - (mx_i - b)$

A variação da função de perda será:

$\partial L = \frac{\partial L}{\partial m} \partial m + \frac{\partial L}{\partial b} \partial b =\begin{pmatrix}
  \frac{\partial L}{\partial m}\\
  \frac{\partial L}{\partial b}
\end{pmatrix}  \begin{pmatrix}
  \partial m\\
  \partial b
\end{pmatrix}= \nabla L(m,b) \cdot \partial\vec{p}$

Encontramos nosso Gradiente Descendente. Se quisermos minimizar o erro, seguiremos a direção inversa da maior variação. Agora temos a direção e as variáveis ​​que precisamos otimizar, mas como? A ideia é começar com ambas como valores aleatórios (zero, para facilitar) e atualizar seus valores a cada rodada de aprendizado, que chamamos de *Época*, e o quanto consideraremos a direção do gradiente, as *Taxas de Aprendizado*. Para prosseguir, precisamos entender como as *Taxas de Aprendizagem (LR)* e as *Épocas* funcionam.

Imagine que nossas variáveis, começando com zero, serão atualizadas da seguinte forma:

$m_{i+1} = m_i - \alpha \nabla L(m,b)$

$b_{i+1} = b_i - \alpha \nabla L(m,b)$,

onde a variável atualizada é a anterior, menos a nossa *Taxa de Aprendizagem* multiplicada pelo Gradiente Descendente.

Se, a cada atualização das variáveis, dermos muito crédito à direção decrescente, podemos ter uma convergência rápida, mas corremos o risco de o resultado divergir. Por outro lado, se não dermos crédito suficiente, a convergência pode ser muito lenta. Esse crédito que damos na direção indicada por $\nabla L(m,b)$ é chamado de Taxa de Aprendizagem e é simbolizado por $\alpha$.

Agora que temos a estrutura completa, precisamos de uma frequência de atualização, e é aí que entra a nossa *Época*. Em geral, temos três formatos:

- Gradiente Descendente em Lote: aqui atualizamos $\nabla L(m,b)$ usando todos os dados disponíveis. É uma abordagem segura, percorremos todos os dados e usamos o vetor de resultado como direção.
- Gradiente Descendente Estocástico: aqui atualizamos $\nabla L(m,b)$ para cada ponto dos nossos dados. É um passo caótico, damos um passo a cada ponto.
- Gradiente Descendente em Mini-Lotes: aqui atualizamos $\nabla L(m,b)$ para cada N/lote em nossos dados. É um passo moderado, damos um passo para cada grupo de dados que observamos.

Na lógica completa, temos:

Nosso GD nos fornece uma direção para o ponto de maior diminuição, nosso LR nos dá o tamanho do nosso passo nessa direção e nossa época nos dá quantos passos daremos e com base em quantos dados.

NOTAS: 

1) *Gradiente Descendente*

Para maximizar a variação, $||d\vec{p}||$ deve ser igual a 1, de modo que capturemos toda a variação. Usando a desigualdade de Cauchy-Schwarz, temos:

$\nabla L \cdot d\vec{p} \leq ||\nabla L||\cdot ||d\vec{p}||$

seguindo para $||d\vec{p}||$ = 1, então

$\vec{\nabla} L \cdot d\vec{p} \leq ||\nabla L||$, logo


$d\vec{p} \leq \frac{\vec{\nabla} L}{||\nabla L||}$

Assim, vemos agora que, para seguir a direção de maior aumento, devemos seguir a mesma direção do gradiente $\vec{\nabla} L(m,b)$.

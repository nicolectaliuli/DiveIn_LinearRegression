EN

What is Linear Regression? Linear Regression is a statistical and supervised learning approach that fits a linear function to a dataset. The goal is to minimize the Cost Function, usually the Mean Squared Error (MSE), which measures the difference between the actual values ​​and the model's predictions. In Machine Learning, we seek the relationship between a target variable ($y$) and one or more exploratory features ($x$). The fundamental structure is the Affine Function:

$f(x) = \sum_{i=0}^{N} a_ix_i$

To find the best parameters ($m$ and $b$), we use a Loss Function that quantifies the residual error:

$J(m,b) = \frac{1}{N}\sum_{i=0}^{N}[y_i - (mx_i+b)]^2$

where:

- $y_i$: is the observed actual value
- $m*x_i+b$: is the predicted value

How do linear regression models work?

The model adjustment is conducted using the Gradient Descent algorithm. The idea is to observe how the total error varies in relation to small changes in the parameters $m$ (slope) and $b$ (intercept). To do this, we calculate the gradient vector ($\nabla J$), composed of the partial derivatives of the cost function:

Partial derivative with respect to *m*

$\frac{\partial J}{\partial m} = \frac{-2}{n}\sum_{i=0}^{N} x_n*[y_i - (mx_i - b)]$

Partial derivative with respect to *b*

$\frac{\partial J}{\partial b} = \frac{-2}{n}\sum_{i=0}^{N} y_i - (mx_i - b)$

The variation of the loss function will be:

$\partial J = \frac{\partial J}{\partial m} \partial m + \frac{\partial J}{\partial b} \partial b =\begin{pmatrix}

\frac{\partial J}{\partial m}\\
\frac{\partial J}{\partial b}
\end{pmatrix} \begin{pmatrix}

\partial m\\

\partial b
\end{pmatrix}= \nabla J(m,b) \cdot \partial\vec{p}$

Optimization and Hyperparameters: Gradient Descent points in the direction of greatest error growth. Since we want to minimize the error, we update the parameters in the opposite direction:

$m_{new} = m_{old} - \alpha \frac{\partial J}{\partial m}$

$b_{new} = b_{old} - \alpha \frac{\partial J}{\partial b}$

Here, two crucial concepts come into play:

- Learning Rate ($\alpha$): Defines the size of the "step". If it is too high, the model may diverge and never find the minimum. If it is too low, the training will be excessively slow.

- Epochs: Represent how many times the algorithm will see the entire dataset.

Update Strategies

There are three main ways to apply Gradient Descent based on the update frequency:

- Batch Gradient Descent: Calculates the error for all examples before updating the parameters. It is stable, but slow for large volumes of data.

- Stochastic Gradient Descent (SGD): Updates parameters for each individual sample. It's fast and helps avoid local minima, but the path to convergence is "noisy".

- Mini-Batch Gradient Descent: The ideal balance. Updates parameters after observing a small group (e.g., 32 or 64 samples). It's the industry standard.

Technical Note: Why does the Gradient point towards the greatest variation?

To demonstrate that the gradient $\nabla J$ indicates the direction of maximum growth, we analyze the variation of the function $dJ = \nabla J \cdot d\vec{p}$, assuming a unit norm shift ($\|d\vec{p}\| = 1$), applying the Cauchy-Schwarz inequality:
To maximize the variation, $||d\vec{p}||$ must be equal to 1, so that we capture all the variation coming from $\nabla J$. Using the Cauchy-Schwarz inequality, we have:

$\nabla J \cdot d\vec{p} \leq ||\nabla J||\cdot ||d\vec{p}||$

Following this, $||d\vec{p}||$ = 1, then

$\vec{\nabla} J \cdot d\vec{p} \leq ||\nabla J||$, therefore the maximum variation will be the norm of $\nabla J$,

$d\vec{p} \leq \frac{\vec{\nabla} J}{||\nabla J||}$

Thus, we now see that, to follow the direction of greatest increase, we must follow the same direction as the gradient $\vec{\nabla} J(m,b)$.

O que é Regressão Linear?A Regressão Linear é uma abordagem estatística e de aprendizado supervisionado que ajusta uma função linear a um conjunto de dados. O objetivo é minimizar a Função de Custo, geralmente o Erro Quadrático Médio (MSE), que mede a diferença entre os valores reais e as previsões do modelo.Em Aprendizado de Máquina, buscamos a relação entre uma variável alvo ($y$) e uma ou mais características exploratórias ($x$). A estrutura fundamental é a Função Afim:


$f(x) = \sum_{i=0}^{N} a_ix_i$


Para encontrar os melhores parâmetros ($m$ e $b$), utilizamos uma Função de Perda que quantifica o erro residual:


$J(m,b) = \frac{1}{N}\sum_{i=0}^{N}[y_i - (mx_i+b)]^2$

onde:
- $y_i$: é o valor real observado
- $m*x_i+b$: é o valor previsto


Como funcionam os modelos de regressão linear?

O ajuste do modelo é conduzido pelo algoritmo de Gradiente Descendente. A ideia é observar como o erro total varia em relação a pequenas mudanças nos parâmetros $m$ (inclinação) e $b$ (intercepto). Para isso, calculamos o vetor gradiente ($\nabla J$), composto pelas derivadas parciais da função de custo:

Derivada parcial em relação a *m*

$\frac{\partial J}{\partial m} = \frac{-2}{n}\sum_{i=0}^{N} x_n*[y_i - (mx_i - b)]$

Derivada Parcial em relação a *b*

$\frac{\partial J}{\partial b} = \frac{-2}{n}\sum_{i=0}^{N} y_i - (mx_i - b)$

A variação da função de perda será:

$\partial J = \frac{\partial J}{\partial m} \partial m + \frac{\partial J}{\partial b} \partial b =\begin{pmatrix}
  \frac{\partial J}{\partial m}\\
  \frac{\partial J}{\partial b}
\end{pmatrix}  \begin{pmatrix}
  \partial m\\
  \partial b
\end{pmatrix}= \nabla J(m,b) \cdot \partial\vec{p}$

Otimização e HiperparâmetrosO Gradiente Descendente aponta para a direção de maior crescimento do erro. Como queremos minimizar o erro, atualizamos os parâmetros na direção oposta:


$m_{novo} = m_{antigo} - \alpha \frac{\partial J}{\partial m}$


$b_{novo} = b_{antigo} - \alpha \frac{\partial J}{\partial b}$

Aqui entram dois conceitos cruciais:
- Taxa de Aprendizagem ($\alpha$): Define o tamanho do "passo". Se for muito alta, o modelo pode divergir e nunca encontrar o mínimo. Se for muito baixa, o treinamento será excessivamente lento.
- Épocas: Representam quantas vezes o algoritmo verá todo o conjunto de dados.

Estratégias de Atualização


Existem três formas principais de aplicar o Gradiente Descendente baseadas na frequência de atualização:


- Batch Gradient Descent (Lote): Calcula o erro para todos os exemplos antes de atualizar os parâmetros. É estável, mas lento para grandes volumes de dados.

- Stochastic Gradient Descent (SGD): Atualiza os parâmetros para cada exemplo individual. É rápido e ajuda a evitar mínimos locais, mas o caminho até a convergência é "ruidoso".

- Mini-Batch Gradient Descent: O equilíbrio ideal. Atualiza os parâmetros após observar um pequeno grupo (ex: 32 ou 64 amostras). É o padrão na indústria.

$\nabla J \cdot d\vec{p} \leq ||\nabla J||\cdot ||d\vec{p}||$

seguindo para $||d\vec{p}||$ = 1, então

$\vec{\nabla} J \cdot d\vec{p} \leq ||\nabla J||$, logo a variação máxima será a norma de $\nabla J$,


$d\vec{p} \leq \frac{\vec{\nabla} J}{||\nabla J||}$

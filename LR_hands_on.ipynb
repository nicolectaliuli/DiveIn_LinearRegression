Linear Regression Hands On

```python
#Logic Implementation (simple structure, one feature, one target variable)
import pandas as pd

def loss_function(m,b,data):
    total_loss = 0
    for i in range(len(data)):
        x = data[i,0]
        y = data[i,1]
        total_loss += (y - (m*x + b)) ** 2
    return total_loss/len(data)

def batch_gradient_descent(m,b,data,learning_rate):
    m_gradient = 0
    b_gradient = 0
    N = len(data)
    
    for i in range(N):
        x = data[i,0]
        y = data[i,1]
        m_gradient += -(2/N) * x * (y - (m*x + b)) #partial derivative with respect to m
        b_gradient += -(2/N) * (y - (m*x + b)) #partial derivative with respect to b
    
    m -= learning_rate * m_gradient #update m
    b -= learning_rate * b_gradient #update b
    
    return m, b

def stochastic_gradient_descent(m,b,data,learning_rate):
    m_gradient = 0
    b_gradient = 0
    N = len(data)
    
    for i in range(N):
        x = data[i,0]
        y = data[i,1]
        m_gradient += -(2/N) * x * (y - (m*x + b)) #partial derivative with respect to m
        b_gradient += -(2/N) * (y - (m*x + b)) #partial derivative with respect to b
        m -= learning_rate * m_gradient #update m
        b -= learning_rate * b_gradient #update b
    
    return m, b

def mini_batch_gradient_descent(m,b,data,learning_rate, batch_size):
    N = len(data)
    
    for i in range(0, N, batch_size): #loop through mini-batches
        m_gradient = 0
        b_gradient = 0
        batch = data[i:i+batch_size]
        k = len(batch)

        for j in range(k): #loop through each data point in the mini-batch
            x = batch[j,0]
            y = batch[j,1]
            m_gradient += -(2/N) * x * (y - (m*x + b)) #partial derivative with respect to m
            b_gradient += -(2/N) * (y - (m*x + b)) #partial derivative with respect to b

        m -= learning_rate * m_gradient #update m
        b -= learning_rate * b_gradient #update b
    
    return m, b

def train(data, learning_rate=0.01, epochs=1000, method='batch'):
    m = 0 #initial slope
    b = 0 #initial intercept
    
    for epoch in range(epochs):
        if method == 'batch':
            m, b = batch_gradient_descent(m, b, data, learning_rate)
        elif method == 'stochastic':
            m, b = stochastic_gradient_descent(m, b, data, learning_rate)
        elif method == 'mini_batch':
            m, b = mini_batch_gradient_descent(m, b, data, learning_rate, batch_size=2)
        if epoch % 100 == 0: #print loss every 100 epochs
            print(f'Epoch {epoch}, Loss: {loss_function(m, b, data)}')
    
    return m, b

# Example usage
if __name__ == "__main__":
    # Sample data: [feature, target]
    data = pd.DataFrame({
        'feature': [1, 2, 3, 4, 5],
        'target': [2, 3, 5, 7, 11]
    }).values
    
    m, b = train(data, learning_rate=0.01, epochs=1000, method='stochastic')
    print(f'Trained parameters: m = {m}, b = {b}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

#Data Creation
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
data = np.hstack((X, y))

#Functions
def loss_function(m, b, data):
    X, y = data[:, 0], data[:, 1]
    return np.mean((y - (m * X + b))**2)

def batch_gradient_descent(m, b, data, lr):
    X, y = data[:, 0], data[:, 1]
    N = len(data)
    error = y - (m * X + b)
    m_grad = -(2/N) * np.sum(X * error)
    b_grad = -(2/N) * np.sum(error)
    return m - lr * m_grad, b - lr * b_grad

def stochastic_gradient_descent(m, b, data, lr):
    idx = np.random.randint(len(data))
    xi, yi = data[idx, 0], data[idx, 1]
    error = yi - (m * xi + b)
    m_grad = -2 * xi * error
    b_grad = -2 * error
    return m - lr * m_grad, b - lr * b_grad

def mini_batch_gradient_descent(m,b,data,learning_rate, batch_size=10):
    N = len(data)
    
    for i in range(0, N, batch_size): #loop through mini-batches
        m_gradient = 0
        b_gradient = 0
        batch = data[i:i+batch_size]
        k = len(batch)

        for j in range(k): #loop through each data point in the mini-batch
            x = batch[j,0]
            y = batch[j,1]
            m_gradient += -(2/N) * x * (y - (m*x + b)) #partial derivative with respect to m
            b_gradient += -(2/N) * (y - (m*x + b)) #partial derivative with respect to b

        m -= learning_rate * m_gradient #update m
        b -= learning_rate * b_gradient #update b
    
    return m, b

def train(data, learning_rate=0.05, epochs=100, method='batch'):
    m, b = -1.0, 5.0 #Initial parameters
    theta_history = []
    cost_history = []
    
    for epoch in range(epochs):
        if method == 'batch':
            m, b = batch_gradient_descent(m, b, data, learning_rate)
        elif method == 'stochastic':
            m, b = stochastic_gradient_descent(m, b, data, learning_rate)
        elif method == 'mini_batch':
            m, b = mini_batch_gradient_descent(m, b, data, learning_rate)
        
        #Store history for visualization
        theta_history.append((m, b))
        cost_history.append(loss_function(m, b, data))
            
    return np.array(theta_history), np.array(cost_history)

#Train the model
#Choice of method: 'batch' or 'stochastic' or 'mini_batch'
metodo_escolhido = 'stochastic' 
n_iterations = 100
t_history, c_history = train(data, learning_rate=0.1, epochs=n_iterations, method=metodo_escolhido)

#Surface for Gradient Descent
m_range = np.linspace(-2, 7, 50)
b_range = np.linspace(-1, 12, 50)
M, B = np.meshgrid(m_range, b_range)
Z = np.array([loss_function(mi, bi, data) for mi, bi in zip(np.ravel(M), np.ravel(B))]).reshape(M.shape)

#Animations
plt.ioff()
fig = plt.figure(figsize=(18, 6))
ax1 = fig.add_subplot(131)
ax2 = fig.add_subplot(132)
ax3 = fig.add_subplot(133, projection='3d')

def animate(i):
    ax1.clear(); ax2.clear(); ax3.clear()
    curr_m, curr_b = t_history[i]
    
    #Plot 1: Linear Fit
    ax1.scatter(X, y, alpha=0.5)
    ax1.plot([0, 2], [curr_b, curr_m*2 + curr_b], color='red', lw=3)
    ax1.set_title(f"Method: {metodo_escolhido.upper()}\nIterations: {i}")
    ax1.set_ylim(0, 15)

    #Plot 2: Cost Evolution
    ax2.plot(range(i+1), c_history[:i+1], color='purple', marker='o')
    ax2.set_xlim(0, n_iterations); ax2.set_ylim(0, max(c_history))
    ax2.set_ylabel("MSE")

    #Plot 3: Gradient Descent Evolution
    ax3.plot_wireframe(M, B, Z, color='black', alpha=0.1)
    ax3.plot(t_history[:i+1, 0], t_history[:i+1, 1], c_history[:i+1], color='red', lw=2, marker = "o")
    ax3.scatter(curr_m, curr_b, c_history[i], color='red', s=50)
    ax3.view_init(elev=30, azim=-120)
    ax3.set_title("Path in Parameter Space (m, b)")
ani = FuncAnimation(fig, animate, frames=n_iterations, interval=100)
HTML(ani.to_jshtml())
```
